# Runge-Kutta 1 (RK1 / Euler Method) for Ordinary Differential Equation

```python
def f(x, y):
    return x + y

x = float(input("Enter initial x: "))
y = float(input("Enter initial y: "))
h = float(input("Enter step size h: "))
xn = float(input("Enter final x: "))

print("\n x\t\t y")

while x <= xn:
    print(f"{x:.2f}\t\t{y:.6f}")
    y = y + h * f(x, y)
    x = x + h
```

# Runge-Kutta 4 (RK4) for Ordinary Differential Equation

```python
def f(x, y):
    return x + y

x = float(input("Enter initial x: "))
y = float(input("Enter initial y: "))
h = float(input("Enter step size h: "))
xn = float(input("Enter final x: "))

print("\n x\t\t y")

while x <= xn:
    print(f"{x:.2f}\t\t{y:.6f}")

    k1 = h * f(x, y)
    k2 = h * f(x + h/2, y + k1/2)
    k3 = h * f(x + h/2, y + k2/2)
    k4 = h * f(x + h, y + k3)

    y = y + (k1 + 2*k2 + 2*k3 + k4) / 6
    x = x + h
```

# Simultaneous Runge-Kutta 1 (RK1 / Euler Method)

```python
def f1(x, y, z):
    return z

def f2(x, y, z):
    return -y

x = float(input("Enter initial x: "))
y = float(input("Enter initial y: "))
z = float(input("Enter initial z: "))
h = float(input("Enter step size h: "))
xn = float(input("Enter final x: "))

print("\n x\t\t y\t\t z")

while x <= xn:
    print(f"{x:.2f}\t\t{y:.6f}\t{z:.6f}")

    y_new = y + h * f1(x, y, z)
    z_new = z + h * f2(x, y, z)

    y = y_new
    z = z_new
    x = x + h
```

# Simultaneous Runge-Kutta 4 (RK4 Method)

```python
def f1(x, y, z):
    return z

def f2(x, y, z):
    return -y

x = float(input("Enter initial x: "))
y = float(input("Enter initial y: "))
z = float(input("Enter initial z: "))
h = float(input("Enter step size h: "))
xn = float(input("Enter final x: "))

print("\n x\t\t y\t\t z")

while x <= xn:
    print(f"{x:.2f}\t\t{y:.6f}\t{z:.6f}")

    k1y = h * f1(x, y, z)
    k1z = h * f2(x, y, z)

    k2y = h * f1(x + h/2, y + k1y/2, z + k1z/2)
    k2z = h * f2(x + h/2, y + k1y/2, z + k1z/2)

    k3y = h * f1(x + h/2, y + k2y/2, z + k2z/2)
    k3z = h * f2(x + h/2, y + k2y/2, z + k2z/2)

    k4y = h * f1(x + h, y + k3y, z + k3z)
    k4z = h * f2(x + h, y + k3y, z + k3z)

    y = y + (k1y + 2*k2y + 2*k3y + k4y) / 6
    z = z + (k1z + 2*k2z + 2*k3z + k4z) / 6

    x = x + h
```
