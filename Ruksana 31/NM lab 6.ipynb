 ## 1. RK-2 Ordinary Differential Equation
Find y at x=0.2

dx\dy=x+y

#### Given:
x0=0

y0=1

h=0.2

```python
def f(x, y):
    return x + y

x = 0
y = 1
h = 0.2

k1 = h * f(x, y)
k2 = h * f(x+h, y+k1)

y = y + (k1+k2)/2

print("Value of y =", y)
```

## 2. RK-4 Ordinary Differential Equation
Find y at x=0.2

dx\dy=x+y

#### Given:

x0=0

y0=1

h=0.2

```python
def f(x, y):
    return x + y

x = 0
y = 1
h = 0.2

k1 = h*f(x,y)
k2 = h*f(x+h/2, y+k1/2)
k3 = h*f(x+h/2, y+k2/2)
k4 = h*f(x+h, y+k3)

y = y + (k1+2*k2+2*k3+k4)/6

print("Value of y =", y)
```

## 3.RK-2 Simultaneous Differential Equation
dx\dy=z

dx\dz=x−y

#### Given:

x=0
y=1
z=0
h=0.1

```python
def f(x,y,z):
    return z

def g(x,y,z):
    return x-y


x = 0
y = 1
z = 0
h = 0.1


k1 = h*f(x,y,z)
l1 = h*g(x,y,z)

k2 = h*f(x+h, y+k1, z+l1)
l2 = h*g(x+h, y+k1, z+l1)


y = y + (k1+k2)/2
z = z + (l1+l2)/2


print("Value of y =", y)
print("Value of z =", z)
```

## 4. RK-4 Simultaneous Differential Equation
dx\dy=z

dx\dz=x−y

#### Given:

x=0

y=1

z=0

h=0.1

```python
def f(x,y,z):
    return z

def g(x,y,z):
    return x-y


x = 0
y = 1
z = 0
h = 0.1


k1 = h*f(x,y,z)
l1 = h*g(x,y,z)


k2 = h*f(x+h/2, y+k1/2, z+l1/2)
l2 = h*g(x+h/2, y+k1/2, z+l1/2)


k3 = h*f(x+h/2, y+k2/2, z+l2/2)
l3 = h*g(x+h/2, y+k2/2, z+l2/2)


k4 = h*f(x+h, y+k3, z+l3)
l4 = h*g(x+h, y+k3, z+l3)


y = y + (k1+2*k2+2*k3+k4)/6
z = z + (l1+2*l2+2*l3+l4)/6


print("Value of y =", y)
print("Value of z =", z)
```
